# La Ley de los Grandes Números y el Teorema del Límite Central


## Ley de los Grandes Números (LGN)

- **Jacob Bernoulli (1730):** formuló la Ley de los Grandes Números.
- **Andréi Kolmogórov (1933):** realizó su formalización matemática.

> **Ley de los Grandes Números (LGN):**
>
> Establece que el **promedio muestral** converge (en algún sentido) al **valor esperado** conforme el tamaño de la muestra aumenta.

Es decir,

$$
\bar{X}_n \longrightarrow \mathbb{E}[X]
\quad \text{cuando } n \to \infty.
$$

---

## Teorema del Límite Central (TLC)

Inicialmente se utilizó para aproximar una distribución **binomial** mediante una **distribución normal**.

### Desarrollo histórico

- **Abraham de Moivre (1733):** primera aproximación de una binomial por una normal.
- **Pierre-Simon Laplace (1812):** popularizó y extendió el teorema.
- **Aleksandr Lyapunov (1901):** generalizó el resultado mediante condiciones más amplias.


## Ejemplo

Lanzamos una moneda y sea

$$
X=\text{el número de soles}.
$$

Se realiza el experimento **5000 veces**, por lo que

$$
X \sim \operatorname{Bin}(5000,\tfrac{1}{2}).
$$

Podemos calcular la probabilidad

$$
P(3500 \leq X \leq 4500)
=
\sum_{i=3500}^{4500}
\binom{5000}{i}
\left(\frac{1}{2}\right)^i
\left(1-\frac{1}{2}\right)^{5000-i}.
$$

Sin embargo, esta suma es muy complicada de evaluar directamente, por lo que utilizamos el **Teorema del Límite Central**.

Calculamos primero la esperanza y la varianza:

$$
\mathbb{E}[X]=np=5000\left(\frac{1}{2}\right)=2500,
$$

$$
\operatorname{Var}(X)=np(1-p)
=5000\left(\frac{1}{2}\right)\left(\frac{1}{2}\right)
=1250.
$$

Estandarizamos la variable aleatoria:

$$
Z=\frac{X-\mathbb{E}[X]}{\sqrt{\operatorname{Var}(X)}}
\sim \mathcal{N}(0,1).
$$

Entonces,

$$
P(3500\leq X\leq4500)
=
P\left(
\frac{3500-2500}{\sqrt{1250}}
\leq
Z
\leq
\frac{4500-2500}{\sqrt{1250}}
\right).
$$

Calculando los valores:

$$
P(3500\leq X\leq4500)
=
P(28.28\leq Z\leq56.57)
\approx 0.
$$

> **Interpretación:** La probabilidad de obtener entre 3500 y 4500 soles en 5000 lanzamientos de una moneda justa es prácticamente nula, ya que dicho intervalo se encuentra muy alejado del valor esperado de 2500.

#NOTA

| Variable aleatoria | Vector aleatorio |
|:------------------|:-----------------|
| $X:\Omega\to\mathbb{R}$ | $(X,Y):\Omega\to\mathbb{R}^2$ |
| Produce un número real | Produce un par ordenado |
| Se representa sobre una recta | Se representa sobre un plano |
| $F_X(x)=P(X\le x)$ | $F_{X,Y}(x,y)=P(X\le x,\,Y\le y)$ |
| Acumula probabilidad hasta $x$ | Acumula probabilidad en un rectángulo del plano |

# Vector Aleatorio

$$
\mathbf{X}=(X_1,\ldots,X_n):\Omega\rightarrow\mathbb{R}^n
$$

Asigna a cada resultado del experimento un vector de \(n\) variables aleatorias.

## Discreto

$$
f_{X,Y}(x,y)=P(X=x,\;Y=y)
$$

 $$f(x,y)\ge0$$

 $$\sum_x\sum_y f(x,y)=1$$

## Continuo

$$
f_{X,Y}(x,y)
$$

$$f(x,y)\ge0$$

 $$\displaystyle\iint_{\mathbb{R}^2}f(x,y)\,dx\,dy=1$$

Las probabilidades se calculan mediante

$$
P\big((X,Y)\in A\big)
=
\iint_A f(x,y)\,dx\,dy.
$$

> Basicamente en el caso **discreto** se **suman** probabilidades,  en el  **continuo** se **integran** densidades.

# Distribuciones Marginales

A partir de la **distribución conjunta** \(f_{X,Y}(x,y)\) se pueden obtener las distribuciones marginales de \(X\) y \(Y\).

## Caso discreto

Las funciones de masa marginales se obtienen sumando sobre la otra variable:

$$
f_X(x)=\sum_y f_{X,Y}(x,y),
$$

$$
f_Y(y)=\sum_x f_{X,Y}(x,y).
$$



## Caso continuo

Las funciones de densidad marginales se obtienen integrando sobre la otra variable:

$$
f_X(x)=\int_{\mathbb{R}} f_{X,Y}(x,y)\,dy,
$$

$$
f_Y(y)=\int_{\mathbb{R}} f_{X,Y}(x,y)\,dx.
$$

> **Idea clave:** Para obtener la distribución marginal de una variable, se "elimina" la otra:
>
> - **Discreto:** se **suma**.
> - **Continuo:** se **integra**.

---

# Independencia

Las variables aleatorias \(X\) y \(Y\) son **independientes** si conocer el valor de una no proporciona información sobre la otra.

## Caso discreto

$$
P(X=x,\;Y=y)=P(X=x)\,P(Y=y).
$$



## Caso continuo

$$
f_{X,Y}(x,y)=f_X(x)\,f_Y(y).
$$

> **Recordatorio:** Esta condición es equivalente a la regla de independencia de eventos:
>
> $$
> P(A\cap B)=P(A)\,P(B).
> $$

#NOTA
>
> - Si quieres obtener \(f_X(x)\), debes **eliminar \(Y\)** (sumando o integrando respecto a \(y\)).
> - Si quieres obtener \(f_Y(y)\), debes **eliminar \(X\)** (sumando o integrando respecto a \(x\)).

# Variables Aleatorias i.i.d.

Sean

$$
X_1,X_2,\ldots,X_n
$$

variables aleatorias.

Se dice que son **independientes** si su distribución conjunta puede escribirse como el producto de sus distribuciones individuales:

$$
f_{X_1,\ldots,X_n}(x_1,\ldots,x_n)
=
\prod_{i=1}^{n}f_{X_i}(x_i).
$$

Si además todas tienen la misma distribución, entonces

$$
E[X_i]=\mu,
\qquad
\operatorname{Var}(X_i)=\sigma^2,
\quad i=1,\ldots,n.
$$

En este caso se dice que las variables son independientes e idénticamente distribuidas.

# Ley de los Grandes Números (LGN)

Sea

$$
S_n=\sum_{i=1}^{n}X_i=X_1+\cdots+X_n.
$$

Como las variables son i.i.d.,

$$
E[S_n]=\sum_{i=1}^{n}E[X_i]=n\mu,
$$

y

$$
\operatorname{Var}(S_n)
=
\sum_{i=1}^{n}\operatorname{Var}(X_i)
=
n\sigma^2.
$$

Definimos el promedio muestral

$$
\bar X_n=\frac{S_n}{n}.
$$

La **Ley de los Grandes Números** establece que

$$
\boxed{
\bar X_n
\longrightarrow
\mu
\qquad (n\rightarrow\infty)
}
$$

Es decir, **el promedio de la muestra converge al valor esperado de la población**.

### Tipos de convergencia

- **LGN débil:** convergencia en probabilidad.

- **LGN fuerte:** convergencia casi segura.

# Teorema del Límite Central (TLC)

Si

$$
X_1,\ldots,X_n
$$

son variables aleatorias i.i.d. con media \(\mu\) y varianza \(\sigma^2\), entonces la suma estandarizada

$$
Z_n
=
\frac{S_n-n\mu}{\sqrt{n\sigma^2}}
=
\frac{S_n-n\mu}{\sigma\sqrt n}
$$

converge en distribución a una normal estándar:

$$
\boxed{
Z_n
\overset{d}{\longrightarrow}
\mathcal N(0,1)
}
$$

Es decir, **aunque la distribución original no sea normal, la suma (o el promedio) de muchas variables independientes se aproxima a una distribución normal**.



## Ejemplos

# 1. Variables Bernoulli

Si

$$
X_i\sim\operatorname{Bern}(p),
$$

entonces

$$
\bar X_n
=
\frac{X_1+\cdots+X_n}{n}
\longrightarrow
p.
$$



# 2. Variables Uniformes

Si

$$
X_i\sim U(0,1),
$$

entonces

$$
\bar X_n
=
\frac{S_n}{n}
\longrightarrow
\frac12.
$$

#NOTA

| Ley de los Grandes Números | Teorema del Límite Central |
|-----------------------------|----------------------------|
| Habla del **promedio muestral** | Habla de la **distribución de la suma o promedio** |
| $$\bar X_n\rightarrow\mu$$ | $$Z_n\rightarrow\mathcal N(0,1)$$ |
| Indica **a qué valor converge** | Indica **cómo converge** |
| Convergencia en probabilidad (o casi segura) | Convergencia en distribución |

# La Ley de los Grandes Números (LGN)

Simularemos los lanzamientos de un dado y calcularemos la media.

Sea el espacio muestral

$$
\Omega=\{1,2,3,4,5,6\}
$$

La media teórica del dado es

$$
E[X]=\frac{1+2+3+4+5+6}{6}=3.5.
$$

```python
# Importamos las librerías

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Fijamos una semilla para reproducibilidad

np.random.seed(42)

# Simulamos 10,000 lanzamientos de un dado

lanzamientos = np.random.randint(1, 7, 10000)

# Calculamos la media acumulada

media_acum = np.cumsum(lanzamientos) / np.arange(1, 10001)

# Creamos un DataFrame

df = pd.DataFrame({
    'Lanzamiento': np.arange(1, 10001),
    'Media Acumulada': media_acum
})

# Mostramos las primeras filas

df
```

```python
# Graficamos la media acumulada

plt.figure(figsize=(10,6))
plt.plot(df['Lanzamiento'],df['Media Acumulada'],label='Media muestral')
# Media teórica
plt.axhline(y=3.5,color='red',linestyle='--',label='Media teórica')
plt.xlabel('Número de lanzamientos')
plt.ylabel('Media acumulada')
plt.title('Simulación de la Ley de los Grandes Números')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

# Teorema Central del Límite (TCL)

```python
# Vamos a hacer una muestra (por ejemplo 100) cada una con 10 observaciones de una binomial

import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt

df = pd.DataFrame() # DataFrame

for i in range(1,101):
    # Muestra aleatoria de binomiales de 10 éxitos
    # Lo realizamos 1000 veces el experimento
    # X_j ~ Bin(1000,0.6), j=1,...,10
    muestra = np.random.binomial(1000,0.6,100)
    col = f"muestra {i}" # Nombre de la columna
    df[col] = muestra # agregar la muestra como nueva columna

df.head(5)

#Calculamos el promedio de cada muestra
df_medias = pd.DataFrame(df.mean(), columns=["Media de la muestra"])

# Visualizar la distribución de las medias
sns.distplot(df_medias)
plt.title("Distribución de medias muestrales")
plt.xlabel("Media")
plt.ylabel("Frecuencia")
plt.show()
```

```python
media_teorica = 1000 * 0.6   # n*p
n_muestras = 100            #  100 columnas
tamano_muestra = 10         # Cada columna tiene 10 datos

plt.figure(figsize=(10,6))
# KDE (linea suavizada)
sns.histplot(df_medias,kde=True, color="seagreen", bins=20)
plt.axvline(x=media_teorica, color='red', linestyle='--', label="Media teórica")
plt.title(f"TCL: Distribución de {n_muestras} medias (n={tamano_muestra})")
plt.xlabel("Valor de la media")
plt.ylabel("Frecuencia")
plt.legend()
plt.show()
```

# Fundamentos de la Estadística Inferencial

La **estadística inferencial** busca obtener conclusiones sobre una **población** a partir de una **muestra**.

Para que las inferencias sean válidas, la muestra debe ser:

- Un conjunto de datos.
- Aleatoria.
- Representativa de la población.

---

## Población

La población posee **parámetros fijos**, generalmente desconocidos, como:

- Media poblacional

$$
\mu
$$

- Varianza poblacional

$$
\sigma^2
$$

Estos parámetros describen las características de toda la población.



## Muestra aleatoria

Sea

$$
X_1,X_2,\ldots,X_n
$$

una muestra aleatoria de variables **i.i.d.** (*independientes e idénticamente distribuidas*).

A partir de la muestra se calculan estadísticos, por ejemplo:

### Media muestral

$$
\bar X
=
\frac1n\sum_{i=1}^{n}X_i.
$$

### Varianza muestral

$$
S^2
=
\frac1{n-1}
\sum_{i=1}^{n}(X_i-\bar X)^2.
$$

#NOTA
Los **parámetros** pertenecen a la población y son constantes; los **estadísticos** se calculan con la muestra y son variables aleatorias.

El objetivo de la inferencia estadística es utilizar los estadísticos muestrales para aproximar los parámetros poblacionales.

# Estimadores

Un **estimador** es una función de la muestra aleatoria que se utiliza para aproximar un parámetro desconocido de la población.

Si

$$
X_1,\ldots,X_n
$$

es una muestra aleatoria, un estimador se escribe como

$$
\hat{\theta}
=
T(X_1,\ldots,X_n),
$$

donde

- \(\theta\) es el parámetro poblacional.
- \(\hat\theta\) es el estimador de dicho parámetro.



## Ejemplos

La media muestral estima la media poblacional:

$$
\hat\mu=\bar X.
$$

La varianza muestral estima la varianza poblacional:

$$
\hat{\sigma}^2=S^2.
$$

Un estimador no tiene por qué ser único. Por ejemplo,

$$
T(X_1,X_2)
=
\frac{X_1+X_2}{2},
$$

o incluso

$$
T(X_1,X_2)
=
\frac{3X_1+X_2}{4},
$$

también son funciones de la muestra.

La diferencia entre ellos radica en sus propiedades estadísticas.



## Espacio parametral

El conjunto de todos los posibles valores que puede tomar un parámetro se denomina **espacio parametral**:

$$
\Theta
=
\{\theta:\theta\text{ es un parámetro del modelo}\}.
$$

#NOTA
 El estimador es una **variable aleatoria**, mientras que el parámetro es un **valor fijo (desconocido)**.

# Propiedades de un Buen Estimador

## 1. Insesgado

Un estimador es **insesgado** si su valor esperado coincide con el parámetro verdadero.

$$
E[\hat\theta]=\theta.
$$

Es decir, en promedio estima correctamente el parámetro.


## 2. Consistente

Un estimador es **consistente** si converge al parámetro cuando el tamaño de la muestra aumenta.

$$
\hat\theta_n
\longrightarrow
\theta,
\qquad
n\to\infty.
$$

Esto significa que, con muestras grandes, el estimador se aproxima al valor real.



## 3. Eficiente

Entre dos estimadores insesgados, es preferible aquel cuya varianza sea menor.

En general,

$$
\operatorname{Var}(\hat\theta)
\text{ lo más pequeña posible.}
$$

Una menor varianza implica estimaciones más precisas.


## 4. Suficiente

Un estimador es **suficiente** si contiene toda la información de la muestra referente al parámetro que se desea estimar.

En otras palabras, conocer el estimador es tan informativo como conocer toda la muestra respecto al parámetro.



## 5. Completo

Un estimador suficiente es **completo** si no existe otro estimador insesgado basado en la misma información que proporcione una mejor estimación.

La completitud es una propiedad importante porque garantiza la existencia de un **estimador insesgado de mínima varianza (UMVU)**.


## Nota

| **Propiedad** | **Significado** |
|:-------------|:----------------|
| **Insesgado** | $$E[\hat{\theta}]=\theta$$ |
| **Consistente** | $$\hat{\theta}\rightarrow\theta,\qquad n\rightarrow\infty$$ |
| **Eficiente** | Tiene la menor varianza posible. |
| **Suficiente** | Aprovecha toda la información de la muestra. |
| **Completo** | No existe otro estimador insesgado mejor basado en la misma información. |

# Método de los Momentos

El **método de los momentos** consiste en construir estimadores igualando los **momentos poblacionales** con los **momentos muestrales**.

## Momento poblacional

Si \(X\) es una variable aleatoria, el momento poblacional de orden \(k\) es

$$
E[X^k].
$$

En particular,

- Primer momento:

$$
E[X]=\mu.
$$

- Segundo momento:

$$
E[X^2].
$$


## Momento muestral

El momento muestral de orden \(k\) es

$$
m_k
=
\frac1n\sum_{i=1}^{n}X_i^k.
$$

Por ejemplo,

- Primer momento muestral:

$$
m_1=\bar X.
$$

- Segundo momento muestral:

$$
m_2=
\frac1n\sum_{i=1}^{n}X_i^2.
$$



## Idea del método

Si la distribución tiene \(p\) parámetros desconocidos,

1. Se calculan los primeros \(p\) momentos poblacionales.
2. Se igualan con los correspondientes momentos muestrales.
3. Se resuelve el sistema de ecuaciones para obtener los estimadores.

##NOTA
 El número de ecuaciones debe ser igual al número de parámetros desconocidos.

# Ejemplo: Distribución Gamma

Sea

$$
X_1,\ldots,X_n
\sim
\Gamma(\alpha\,β),
$$

donde alpha y beta son parámetros desconocidos.

Se desea encontrar los estimadores

$$
\hat\alpha,
\qquad
\hat\beta
$$

mediante el **método de los momentos**.


Sabemos que

$$
E[X]
=
\frac{\alpha}{\beta},
$$

y

$$
\operatorname{Var}(X)
=
\frac{\alpha}{\beta^2}.
$$

Además,

$$
\operatorname{Var}(X)
=
E[X^2]-E[X]^2,
$$

por lo que

$$
E[X^2]
=
\operatorname{Var}(X)+E[X]^2.
$$

Sustituyendo,

$$
E[X^2]
=
\frac{\alpha}{\beta^2}
+
\left(\frac{\alpha}{\beta}\right)^2
=
\frac{\alpha+\alpha^2}{\beta^2}.
$$

## Igualamos momentos poblacionales y muestrales

Primer momento

$$
E[X]
=
\bar X.
$$

Por tanto,

$$
\frac{\alpha}{\beta}
=
\bar X.
$$

Despejando,

$$
\boxed{
\beta
=
\frac{\alpha}{\bar X}
}
$$


Segundo momento

$$
E[X^2]
=
m_2
=
\frac1n
\sum_{i=1}^{n}X_i^2.
$$

Sustituyendo la expresión anterior para \(\beta\),

$$
m_2
=
\bar X^2
\left(
1+\frac1\alpha
\right).
$$

Despejando,

$$
\frac1\alpha
=
\frac{m_2}{\bar X^2}-1,
$$

por lo que

$$
\boxed{
\hat\alpha
=
\frac{\bar X^2}
{m_2-\bar X^2}
}
$$

# Estimadores por el Método de los Momentos

Los estimadores obtenidos son

$$
\boxed{
\hat\alpha
=
\frac{n\bar X^2}
{\sum_{i=1}^{n}(X_i-\bar X)^2}
}
$$


$$
\boxed{
\hat\beta
=
\frac{n\bar X}
{\sum_{i=1}^{n}(X_i-\bar X)^2}
}
$$

Ambos estimadores dependen únicamente de los datos muestrales \(X_1,\ldots,X_n\), por lo que cumplen la definición de estimador.

#NOTA
El método de los momentos suele ser el primer método para obtener estimadores porque es sencillo. Sin embargo, en muchos modelos se prefiere el Método de Máxima Verosimilitud (MLE), ya que normalmente produce estimadores con mejores propiedades estadísticas.

# Método de Máxima Verosimilitud (MLE)

Sea

$$
X_1,X_2,\ldots,X_n
$$

una muestra aleatoria i.i.d. proveniente de una distribución con función de densidad (o masa de probabilidad)

$$
f(x;\theta),
$$

donde \(\theta\) es un parámetro desconocido que se desea estimar.


## Función de Verosimilitud

La **función de verosimilitud** mide qué tan probable es observar la muestra para un valor dado del parámetro.

Como las observaciones son independientes,

$$
L(\theta)
=
\prod_{i=1}^{n}
f(x_i;\theta).
$$


## Log-verosimilitud

Para facilitar los cálculos, se toma el logaritmo de la verosimilitud.

$$
\ell(\theta)
=
\ln L(\theta)
=
\ln\left(
\prod_{i=1}^{n}f(x_i;\theta)
\right)
=
\sum_{i=1}^{n}
\ln f(x_i;\theta).
$$

El logaritmo transforma productos en sumas, lo que simplifica la derivación.



## Procedimiento para obtener el MLE

1. Construir la función de verosimilitud \(L(\theta)\).

2. Calcular la log-verosimilitud

$$
\ell(\theta)=\ln L(\theta).
$$

3. Derivar respecto al parámetro

$$
\frac{d\ell(\theta)}{d\theta}.
$$

4. Igualar la derivada a cero

$$
\frac{d\ell(\theta)}{d\theta}=0.
$$

5. Resolver para \(\theta\).

La solución obtenida corresponde al **estimador de máxima verosimilitud (MLE)**.

##NOTA
El MLE es el valor del parámetro que hace más probable la muestra observada.

# Ejemplo: Distribución Bernoulli

Sea

$$
X_1,\ldots,X_n
\sim
\operatorname{Bern}(p),
$$

donde

$$
P(X=x)=p^x(1-p)^{1-x},
\qquad x\in\{0,1\}.
$$

Se desea encontrar el estimador MLE de \(p\).


## Paso 1. Función de verosimilitud

Como las observaciones son independientes,

$$
L(p)
=
\prod_{i=1}^{n}
p^{x_i}(1-p)^{1-x_i}.
$$

Agrupando exponentes,

$$
L(p)
=
p^{\sum x_i}
(1-p)^{\,n-\sum x_i}.
$$



Derivando respecto a \(p\),

$$
\frac{d\ell(p)}{dp}
=
\frac{\sum x_i}{p}
-
\frac{n-\sum x_i}{1-p}.
$$

Igualamos a cero,

$$
\frac{\sum x_i}{p}
-
\frac{n-\sum x_i}{1-p}
=
0.
$$

Multiplicando por \(p(1-p)\),

$$
(1-p)\sum x_i
-
p\left(n-\sum x_i\right)
=
0.
$$

Desarrollando,

$$
\sum x_i
-
p\sum x_i
-
np
+
p\sum x_i
=
0.
$$

Finalmente,

$$
\sum x_i=np,
$$

por lo que

$$
\boxed{
\hat p
=
\frac1n
\sum_{i=1}^{n}X_i
=
\bar X.
}
$$

Es decir, el estimador de máxima verosimilitud para \(p\) en una distribución Bernoulli es la media muestral.

## Notas

### ¿Por qué usamos el logaritmo?

Porque

$$
\ln(ab)=\ln(a)+\ln(b),
$$

por lo que convierte productos en sumas y simplifica las derivadas.


### ¿El logaritmo cambia el máximo?

No.

Como el logaritmo es una función estrictamente creciente,

$$
\arg\max L(\theta)
=
\arg\max \ln L(\theta).
$$

Ambas funciones alcanzan su máximo en el mismo valor de \(\theta\).

### ¿Qué representa la verosimilitud?

Una vez observada la muestra,

- los datos \(x_1,\ldots,x_n\) son **constantes**;
- el parámetro \(\theta\) es la **cantidad desconocida**.

La función de verosimilitud se interpreta como una función del parámetro:

$$

### Ejemplo 1.

Lanzamod una moneda 100 veces, sabiendo que se obtienen 55 soles. Encontrar el EMV para la probabilidad $p$ de obtener sol en un solo lanzamiento.

Sea
$X$: el número de soles que se obtienen al lanzar una moneda
se tiene que $X\sim \text{Bin}(100,p)$, entonces

$$ f(x;p) = \binom{100}{55}p^{55}(1-p)^{100-55}$$

```python
from scipy.stats import binom
import numpy as np

# Vamos a utilizar la libreria Sympy para matemática símbolica
import sympy

# importa la variable símbolica x
from sympy.abc import x
```

https://www.sympy.org/en/features.html

```python
# Vamos a definir a p como una variable símbolica positiva: 0<p<1
p = sympy.symbols('p', positive= True)

# Definimos la función de verosimilitus
# se puede omitir la constante
f = p**55 * (1-p)**45

# Ahora derivamos f con respecto a p
df = sympy.diff(f,p)

# Hay que igualar la derivada a cero y despejar a p
# df=0 para encontrar p
phat = sympy.solve(df,p)[0] # [1] es la segunda solucion y [0] es la primera

print("El estimador de máxima verosimilitud es", phat)
```

### Ejemplo 2.
Vamos a simular una muestra de tamaño 100 de ceros y unos.

```python
# n=1 : un ensayo de Bernoulli
# p : es la probabilidad de éxito
# size : genera 100 observaciones
r = binom.rvs(n=1, p=0.7, size = 100)
r
```

```python
# Vamos a calcular EMV

# Definimos la función de verosimilitud
f = p**x * (1-p)**(1-x)

# Calculamos el producto de todos los elementos
# subs(x,i) sustituye la variable símbolica x por una valor concreto de la muestra
J = np.prod([f.subs(x,i) for i in r])

# Calculamos el logJ de manera expandida
# sympy.log(J) = LogJ
logJ = sympy.expand_log(sympy.log(J))

# Calculamos da derivada e igualamos a cero
phat = sympy.solve(sympy.diff(logJ,p),p)[0]
print("El estimador de máxima verosimilitud es", phat)
```

```python
# Podemos verificar  las soluciones con:
sol = sympy.solve(sympy.diff(logJ,p),p)
sol
```

Encontrar el estimador MLE con Sympy

```python
import sympy
import numpy as np

theta = sympy.symbols('theta', positive=True)
x = sympy.Symbol('x')
#DEfinimos funcion de  densidad exponencial
f = theta * sympy.exp(-theta*x)
datos = np.array([2,3,1,3,5])
#J es la funcion de verosimilitud
J = np.prod([f.subs(x, i) for i in datos])
logJ = sympy.expand_log(sympy.log(J))
print(logJ)
#Derivamos
dlogJ = sympy.diff(logJ, theta)
print(dlogJ)

theta_hat = sympy.solve(dlogJ, theta)[0]

print("El estimador MLE es:", theta_hat)
print("Valor decimal:", theta_hat.evalf())
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Datos
datos = np.array([2,3,1,3,5])

# Valores de theta
theta = np.linspace(0.05,1,300)

# Log-verosimilitud
logL = len(datos)*np.log(theta) - theta*np.sum(datos)

# EMV
theta_hat = len(datos)/np.sum(datos)

plt.figure(figsize=(7,4))
plt.plot(theta, logL, label="Log-verosimilitud")
plt.axvline(theta_hat, color='red', linestyle='--',
            label=f"MLE = {theta_hat:.4f}")
plt.scatter(theta_hat,
            len(datos)*np.log(theta_hat)-theta_hat*np.sum(datos),
            color='red')

plt.xlabel(r'$\theta$')
plt.ylabel(r'$\ell(\theta)$')
plt.title("Función de Log-Verosimilitud")
plt.legend()
plt.grid(True)
plt.show()
```

# Distribuciones Muestrales

Supongamos que una población tiene:

- Media poblacional

$$
\mu
$$

- Varianza poblacional

$$
\sigma^2.
$$

Se toma una muestra aleatoria

$$
X_1,X_2,\ldots,X_n,
$$

y a partir de ella se calculan estadísticos como:

- Media muestral

$$
\bar X.
$$

- Varianza muestral

$$
S^2.
$$

Las distribuciones de estos estadísticos se conocen como **distribuciones muestrales**.


# Distribución $\chi^2$

La distribución **$\chi^2$**, denotada por

$$
\chi^2,
$$

toma únicamente valores positivos:

$$
\chi^2\in(0,\infty).
$$

## Definición

Si

$$
Z\sim\mathcal N(0,1),
$$

entonces

$$
Z^2
\sim
\chi^2_{(1)}.
$$

Más generalmente, si

$$
Z_1,Z_2,\ldots,Z_n
\sim
\mathcal N(0,1)
$$

son variables aleatorias independientes, entonces

$$
\boxed{
\sum_{i=1}^{n}Z_i^2
\sim
\chi^2_{(n)}.
}
$$


## Grados de libertad

El subíndice

$$
\chi^2_{(n)}
$$

representa los **grados de libertad**, es decir, el número de variables normales estándar independientes cuya suma de cuadrados forma la distribución.


## Aplicación

La distribución $\chi^2$se utiliza principalmente para realizar inferencias sobre la **varianza poblacional** \(\sigma^2\).

Algunas aplicaciones son:

- Construcción de intervalos de confianza para \(\sigma^2\).
- Pruebas de hipótesis sobre la varianza.
- Base para otras distribuciones, como las distribuciones **t** de Student y **F**.

 #NOTA
 La distribución $\chi^2$ surge al sumar los cuadrados de variables normales estándar independientes y es fundamental para estudiar la variabilidad de una población.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chi2, t, f, norm
```

```python
#Distribución  chi cuadrada
x_chi = np.linspace(0,20,500)
plt.figure(figsize=(10,6))

# Damos varios grados de libertad
gl_chi = [1,3,6,10]

for gl in gl_chi:
    plt.plot(x_chi, chi2.pdf(x_chi,gl), label=f'Chi cuadrada (gl={gl})')
plt.title('Distribución Chi cuadrada')
plt.xlabel('x')
plt.ylabel('Densidad de probabilidad')
plt.legend()
plt.grid()
plt.show()
```

# Distribución t de Student

La **distribución t de Student** se utiliza para realizar inferencias sobre la **media poblacional** cuando la desviación estándar poblacional \(\sigma\) es **desconocida**.

En este caso, \(\sigma\) se sustituye por la desviación estándar muestral

$$
S.
$$

## Def:

Sean

$$
Z\sim\mathcal N(0,1)
$$

y

$$
W\sim\chi^2_{(\nu)},
$$

variables aleatorias independientes, donde \(\nu\) representa los grados de libertad.

Entonces,

$$
\boxed{
T
=
\frac{Z}{\sqrt{W/\nu}}
\sim
t_{(\nu)}.
}
$$

## Propiedades

- Es una distribución **simétrica** respecto a cero.
- Tiene **colas más pesadas** que la distribución normal estándar.
- Depende de los **grados de libertad** (\(\nu\)).
- A medida que aumentan los grados de libertad, la distribución se aproxima a una normal estándar.

En particular,

$$
t_{(\nu)}
\longrightarrow
\mathcal N(0,1),
\qquad
\nu\rightarrow\infty.
$$

En la práctica, cuando

$$
n>30,
$$

la distribución \(t\) suele aproximarse muy bien a una distribución normal estándar.


## Aplicaciones

La distribución t de Student se utiliza principalmente para:

- Construir intervalos de confianza para la media poblacional cuando \(\sigma\) es desconocida.
- Realizar pruebas de hipótesis sobre la media.
- Analizar muestras pequeñas provenientes de poblaciones normales.

#NOTA
 Si la desviación estándar poblacional \(\sigma\) es desconocida, se utiliza la distribución **t de Student** en lugar de la distribución normal.

```python
x = np.linspace(-5,5,500)
# Grados de libertad
gl_t = [1,3,10,30]

plt.figure(figsize=(8,4))
for gl in gl_t:
    plt.plot(x,t.pdf(x,gl),label=f't (gl={gl})')
plt.title('Distribución t')
plt.xlabel('x')
plt.ylabel('Densidad de probabilidad')
plt.legend()
plt.grid()
plt.show()
```

# Distribución F

La **distribución F de Fisher-Snedecor** se utiliza para comparar **dos varianzas poblacionales**.

Depende de **dos grados de libertad**:

- \(m\): grados de libertad del numerador.
- \(n\): grados de libertad del denominador.


## Def:

Sean

$$
W_1\sim\chi^2_{(m)}
$$

y

$$
W_2\sim\chi^2_{(n)},
$$

variables aleatorias independientes.

Entonces,

$$
\boxed{
F
=
\frac{W_1/m}{W_2/n}
\sim
F_{(m,n)}.
}
$$

## Propiedades

- Toma únicamente valores positivos:

$$
F>0.
$$

- Depende de dos grados de libertad:

$$
F_{(m,n)}.
$$

- No es una distribución simétrica.

- Presenta asimetría positiva (cola hacia la derecha).

## Aplicaciones

La distribución F se utiliza principalmente para:

- Comparar dos varianzas poblacionales.
- Pruebas de igualdad de varianzas.
- Análisis de Varianza (**ANOVA**).
- Regresión lineal (prueba global del modelo).

#Nota
 La distribución F aparece como el cociente de dos variables Chi-cuadrada independientes divididas entre sus respectivos grados de libertad.

```python
#Distribución F

x = np.linspace(0,5,500)
# Grados de libertad
gl_f = [(1,2),(2,5),(5,10),(10,20)] # (gl1,gl2)

plt.figure(figsize=(8,4))
for d1,d2 in gl_f:
    plt.plot(x,f.pdf(x,d1,d2),label=f'F (gl1={d1},gl2={d2})')
plt.title('Distribución F')
plt.xlabel('x')
plt.ylabel('Densidad de probabilidad')
plt.legend()
plt.grid()
plt.show()
```

# Intervalos de confianza
---

Un intervalo de confianza es un rango de valores que con cierta probabilidad (nivel de confianza) contiene al verdadero parámetro poblacional que queremos estimar.

Sea $X_1,\dots, X_n$ una muestra aleatoria con $X\sim\text{N}(\mu,\sigma^2)$, entonces
$$(num_1 < \mu < num2)$$

En los intervalos de confianza, tenemos el error estándar (SE) el cual mide cuanto varía un estimador.
 - si conocemos $\sigma$, entonces: $SE = \sigma/\sqrt{n}$
 - si NO conocemos $\sigma$, utilizamos $s$ y entonces: $SE = s/\sqrt{n

NOTA: a mayor tamaño de la muestra se tiene que el SE será menor y por lo tanto el intervalo de confianza será más preciso.

**Def** Si $X_1,\dots, X_n$ es una muestra aleatoria con densidad $f(x;\theta)$ y sean $T_1=t_1(X_1,\dots, X_n)$ y $T_2=t_2(X_1,\dots, X_n)$ dos estadísticos talque $T_1 \le T_2$ entonces
$$\mathbb{P}(T_1 \le \theta \le T_2) = \gamma$$
Al intervalo $(T_1,T_2)$ se le llama intervalo del $\gamma\times 100\%$ de confianza.


```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st
```

## 1. Intervalo de confianza para la media de una población normal
Sea $X_1,\dots, X_n$ una muestra aleatoria con $X\sim\text{N}(\mu,\sigma^2)$, entonces
 - $\bar{X_i} \sim \mathcal{N}(\mu, \sigma^2/n)$
 - $Z = \frac{\bar{X_i}-\mu}{\sigma/n} \sim \mathcal{N}(0,1)$
 ### $\sigma$ conocida:

Se tiene que el intervalo del $(1-\alpha)\times 100\%$ de confianza para $\mu$ es:
$$ \bar{X}\pm z_{\alpha/2}\frac{\sigma}{\sqrt{n}}$$
donde
 - $z_{\alpha/2}\frac{\sigma}{\sqrt{n}}$ es el margen de error
 - $z_{\alpha/2}$ es el valor crítico

**Ejemplo 1** Para tratar de estimar la media de consumo por cliente en un restaurante, se reunieron datos de una muestra de tamaño $49$ clientes durante tres semanas.

**a)** Supongamos que la desviación estándar poblacional es de $\$2.5$. ¿Cuál es el error estandar de la media?

**b)** Con un nivel de confianza del $95\%$, ¿cuál es el margen de error?

**c)** Si la media muestral es de $\$22.60$, cual es el intervalo de confianza del $95\%$ para la media poblacional.

```python
# Datos:
n = 49 # número de clientes
sigma = 2.5
media_muestral = 22.60
confianza = 0.95

# a) SE
error_est = sigma / np.sqrt(n)
print('El error estándar es:', error_est)

# b) 1-confianza = 0.95 -> 0.025 en cada cola -> (1-confianza)/2
z = st.norm.ppf(1-(1-confianza)/2)
margen_error = z*error_est
print('El margen de error es:', margen_error)

# c) Intervalo de confianza
limite_inf = media_muestral - margen_error
limite_sup = media_muestral + margen_error
print(f'El intervalo del {confianza*100}% es: ({limite_inf:.2f},{limite_sup:.2f})')
```

Por lo tanto, se tiene que el verdadero consumo poblacional en el restaurante se encuentra entre $\$21.90$ y $\$23.30$

### $\sigma$ desconocida:

Se tiene que el intervalo del $(1-\alpha)\times 100\%$ de confianza para $\mu$ es:
$$ \bar{X}\pm t_{\alpha/2}\frac{\sigma}{\sqrt{n}}$$
donde
 - $t_{\alpha/2}\frac{\sigma}{\sqrt{n}}$ es el margen de error
 - $t_{\alpha/2}$ es el valor crítico

**Ejemplo 2**  Supongamos que tenemos los siguientes datos de una muestra aleatoria de 10 observaciones:
$$45,55,67,68,45,79,98,87,84,82$$

Calcular un intervalo de confianza para la media.

```python
# Datos
datos = [45,55,67,68,45,79,98,87,84,82] # n<= 30
n = len(datos)
confianza = 0.99
gl = n-1

media_muestral = np.mean(datos)
error_est = st.sem(datos) # eror estandar muestral

# Vamos a utilizar la función
# scipy.stats.t.interval(confidence, df, loc, scale)
# confidence = confianza
# df = degree freedom (gl)
# loc = media_muestral
# scale = error_est

intervalo = st.t.interval(confianza, gl, media_muestral, error_est)
print(f'El intervalo de confianza del {confianza*100}%: ({intervalo[0]:.2f},{intervalo[1]:.2f})')
```

## 2. Intervalo de confianza para la varianza de una población normal
---

### $\mu$ desconocida:

Se tiene que el intervalo del $(1-\alpha)\times 100\%$ de confianza para $\sigma^2$ es:
$$ \left( \frac{(n-1)s^2}{\chi^{2}_{1-\alpha/2}}, \frac{(n-1)s^2}{\chi^{2}_{\alpha/2}} \right)$$

**Ejemplo 3** Los artículos de cerámica utilizados sobre velas electricas se rompen con diferentes presiones. Supongamos que los datos provienen de una distribución normal. La resistencia a la ruptura fue medida en una muestra de 100 artículos y el promedio fue de $1750$ con una desviación estándar de $315.8$.

**a)** Estimar con un nivel de confianza del $90\%$ a la media poblacional de la presión de la ruptura.

**b)** Estimar con un nivel de confianza del $90\%$ a la varianza poblacional.

```python
# Datos
n = 100
confianza = 0.90
media_muestral = 1750
desv_est = 315.8
error_est = desv_est / np.sqrt(n)

# a) como n>=30, utilizo la normal
intervalo = st.norm.interval(confianza, media_muestral, error_est)
print(f'El intervalo de confianza del {confianza*100}%: ({intervalo[0]:.2f},{intervalo[1]:.2f})')

# b) IC para la varianza
gl = n-1
alpha = 1-confianza
s2 = desv_est**2 # varianza muestral

# Forma 1. Calculando los cuatiles  de la chi cuadrada
chi_inf = st.chi2.ppf(alpha/2,gl)
chi_sup = st.chi2.ppf(1-alpha/2,gl)
int_var = ((gl*s2) / chi_sup, (gl*s2) / chi_inf)
print(f'El intervalo de confianza del {confianza*100}%: ({int_var[0]:.2f},{int_var[1]:.2f})')

# Forma 2.
chi2_inf, chi2_sup = st.chi2.interval(confianza, gl)
int2_var = ((gl*s2) / chi2_sup, (gl*s2) / chi2_inf)
print(f'El intervalo de confianza del {confianza*100}%: ({int2_var[0]:.2f},{int2_var[1]:.2f})')
```
