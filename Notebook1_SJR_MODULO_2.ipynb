# Introducción: Fundamentos de la Probabilidad

La formulación moderna de la probabilidad fue desarrollada por **Andréi Kolmogórov** y publicada en **1933**.

Su principal aportación fue establecer una **construcción axiomática** de la probabilidad, relacionándola con la **teoría de la medida**, desarrollada previamente por **Émile Borel** y **Henri Lebesgue**.

Gracias a esta formulación, la probabilidad adquirió una base matemática rigurosa.



## Teoría de la medida

La teoría de la medida proporciona las herramientas matemáticas para asignar un "tamaño" o una **medida** a determinados conjuntos.

En probabilidad, esa medida recibe el nombre de **probabilidad**.

Esquema conceptual:

$$
\boxed{
\text{Espacio medible}
\;\longrightarrow\;
\text{Medida}
\;\longrightarrow\;
\text{Probabilidad}
}
$$



## Axiomas de Kolmogórov

La teoría moderna de la probabilidad se basa en tres axiomas fundamentales:

1. **No negatividad**

$$
P(A)\ge0.
$$

2. **Normalización**

$$
P(\Omega)=1.
$$

3. **Aditividad numerable**

Si

$$
A_1,A_2,\ldots
$$

son eventos mutuamente excluyentes,

$$
P\left(\bigcup_{i=1}^{\infty}A_i\right)
=
\sum_{i=1}^{\infty}P(A_i).
$$

##NOTA
Toda la teoría moderna de la probabilidad se construye a partir de estos tres axiomas.

# Experimento Aleatorio

Un **experimento aleatorio** es un experimento que, al repetirse bajo las mismas condiciones, puede producir diferentes resultados y cuyo resultado **no puede predecirse con certeza**.

## Características

- Se realiza bajo las mismas condiciones.
- Puede producir distintos resultados.
- El resultado individual es incierto.
- El conjunto de todos los resultados posibles sí puede describirse.



## Ejemplos

- Lanzar una moneda.
- Lanzar un dado.
- Extraer una carta de una baraja.
- Medir el tiempo de vida de un componente electrónico.

Aunque el resultado de un experimento aleatorio es impredecible, su comportamiento puede describirse mediante modelos probabilísticos.

# Espacio Muestral

El **espacio muestral** (o **espacio de estados**) es el conjunto de **todos los posibles resultados** de un experimento aleatorio.

Se denota por

$$
\Omega.
$$

Cada resultado posible del experimento se denomina **resultado elemental** y se representa por

$$
\omega\in\Omega.
$$


## Ejemplos

### Lanzamiento de una moneda

$$
\Omega=\{\text{Águila},\text{Sol}\}.
$$



### Lanzamiento de un dado

$$
\Omega=\{1,2,3,4,5,6\}.
$$



### Lanzamiento de un dardo sobre un círculo de radio 15 cm

Cada resultado corresponde a un punto del plano.

$$
\Omega=
\left\{
(x,y)\in\mathbb{R}^2:
x^2+y^2\le15^2
\right\}.
$$


### Tiempo de espera

Si Romeo espera a Julieta entre 0 y 1 hora,

$$
\Omega=[0,1].
$$



### Tiempo de vida de una batería

$$
\Omega=[0,\infty).
$$

---

El espacio muestral contiene **todos** los resultados posibles del experimento, independientemente de cuál ocurra finalmente.

# Eventos Aleatorios

Un **evento aleatorio** es un subconjunto del espacio muestral \(\Omega\).

Es decir,

$$
A\subseteq\Omega.
$$

Un evento ocurre cuando el resultado del experimento pertenece al conjunto \(A\).



## Ejemplos

### Lanzamiento de una moneda

Si

$$
\Omega=\{\text{Águila},\text{Sol}\},
$$

el evento

$$
A=\{\text{Sol}\}
$$

representa "obtener sol".



### Lanzamiento de un dado

Si

$$
\Omega=\{1,2,3,4,5,6\},
$$

el evento

$$
A=\{3\}
$$

representa "obtener un 3".



### Tiempo de espera

Si

$$
\Omega=[0,1],
$$

el evento

$$
A=\left[0,\frac14\right]
$$

representa que Julieta llega antes de 15 minutos.

Un evento es un conjunto de resultados para los cuales ocurre una determinada situación de interés.

# ¿Cuántos eventos existen?

El conjunto formado por **todos los subconjuntos** del espacio muestral se denomina **conjunto potencia**.

Se denota por

$$
\mathcal P(\Omega).
$$

Si el espacio muestral tiene

$$
|\Omega|=n
$$

elementos, entonces

$$
\boxed{
|\mathcal P(\Omega)|
=
2^n.
}
$$

Es decir, el número total de eventos posibles es

$$
2^{|\Omega|}.
$$



## Ejemplo

Si

$$
\Omega=\{1,2,3\},
$$

entonces

$$
|\Omega|=3,
$$

y el número de eventos posibles es

$$
2^3=8.
$$

Estos son:

$$
\begin{aligned}
\mathcal P(\Omega)=
\{&
\emptyset,
\{1\},
\{2\},
\{3\},
\{1,2\},
\{1,3\},
\{2,3\},
\Omega
\}.
\end{aligned}
$$

# Eventos de Interés

## Evento seguro

Es el evento que siempre ocurre.

$$
\boxed{\Omega}
$$



## Evento imposible

Es el evento que nunca ocurre.

$$
\boxed{\emptyset}
$$

## Complemento

El complemento de un evento \(A\) contiene todos los resultados donde **no ocurre** \(A\).

Se denota por

$$
A^c.
$$

Ejemplo:

Si

$$
A=\{\text{Sol}\},
$$

entonces

$$
A^c=\{\text{Águila}\}.
$$


## Intersección

El evento

$$
A\cap B
$$

representa que **ocurren simultáneamente** los eventos \(A\) y \(B\).



## Unión

El evento

$$
A\cup B
$$

representa que ocurre \(A\), ocurre \(B\), o ambos.




## Resumen

| **Operación** | **Significado** |
|:-------------:|:----------------|
| <span>$\Omega$</span> | Evento seguro |
| <span>$\emptyset$</span> | Evento imposible |
| <span>$A^{c}$</span> | Complemento |
| <span>$A\cap B$</span> | Ocurren ambos eventos |
| <span>$A\cup B$</span> | Ocurre al menos uno de los eventos |

# σ-Álgebra

Una **σ-álgebra** es una colección de eventos sobre los cuales es posible definir una probabilidad.

Se denota por

$$
\mathcal F.
$$

Debe cumplir las siguientes propiedades.


## 1. Contiene al espacio muestral

$$
\Omega\in\mathcal F.
$$


## 2. Cerrada bajo complemento

Si

$$
A\in\mathcal F,
$$

entonces

$$
A^c\in\mathcal F.
$$


## 3. Cerrada bajo uniones numerables

Si

$$
A_1,A_2,\ldots\in\mathcal F,
$$

entonces

$$
\boxed{
\bigcup_{i=1}^{\infty}A_i
\in
\mathcal F.
}
$$

Como consecuencia,

también es cerrada bajo intersecciones numerables.


## Caso finito

Si el espacio muestral es finito,

una elección común es

$$
\boxed{
\mathcal F=\mathcal P(\Omega),
}
$$

es decir,

la σ-álgebra coincide con el conjunto potencia.

 La σ-álgebra es el conjunto de todos los eventos a los que se les puede asignar una probabilidad.

# Espacio de Probabilidad

Un **espacio de probabilidad** es la terna

$$
(\Omega,\mathcal F,P),
$$

donde:

- \(\Omega\): espacio muestral.
- \(\mathcal F\): σ-álgebra de eventos.
- \(P\): medida de probabilidad.

La función de probabilidad

$$
P:\mathcal F\longrightarrow[0,1]
$$

debe satisfacer los siguientes axiomas.


## Axiomas de Kolmogórov

### 1. No negatividad

Para todo evento \(A\in\mathcal F\),

$$
P(A)\ge0.
$$


### 2. Normalización

La probabilidad del espacio muestral es

$$
P(\Omega)=1.
$$



### 3. Aditividad numerable

Si

$$
A_1,A_2,\ldots
$$

son eventos mutuamente excluyentes (disjuntos),

$$
A_i\cap A_j=\emptyset,
\qquad i\neq j,
$$

entonces

$$
\boxed{
P\left(
\bigcup_{i=1}^{\infty}A_i
\right)
=
\sum_{i=1}^{\infty}
P(A_i).
}
$$
Toda la teoría de la probabilidad moderna se deriva de estos tres axiomas.

# Propiedades de la Probabilidad

A partir de los axiomas se obtienen las siguientes propiedades.

## 1. Evento imposible

$$
P(\emptyset)=0.
$$



## 2. Complemento

$$
P(A^c)=1-P(A).
$$

## 3. Diferencia de eventos

Si

$$
A\subseteq B,
$$

entonces

$$
P(B\setminus A)
=
P(B)-P(A).
$$



## 4. Unión de dos eventos

$$
\boxed{
P(A\cup B)
=
P(A)+P(B)-P(A\cap B).
}
$$

Si \(A\) y \(B\) son disjuntos,

$$
P(A\cup B)=P(A)+P(B).
$$


## 5. Unión de tres eventos

$$
\begin{aligned}
P(A\cup B\cup C)
=&\,
P(A)+P(B)+P(C)\\
&-P(A\cap B)-P(A\cap C)-P(B\cap C)\\
&+P(A\cap B\cap C).
\end{aligned}
$$

Al sumar probabilidades se restan las intersecciones para evitar contar eventos más de una vez.

# Modelo Clásico (Laplace)

El modelo clásico de probabilidad fue propuesto por **Pierre-Simon Laplace** en 1812.

Se basa en el principio de **equiprobabilidad**, es decir, todos los resultados elementales tienen la misma probabilidad.

Si

$$
\Omega=\{\omega_1,\omega_2,\ldots,\omega_n\},
$$

entonces

$$
P(\omega_i)=\frac1n.
$$

La probabilidad de un evento \(A\subseteq\Omega\) es

$$
\boxed{
P(A)
=
\frac{\text{casos favorables}}
{\text{casos posibles}}
=
\frac{|A|}{|\Omega|}.
}
$$

donde

- \(|A|\): número de elementos de \(A\).
- \(|\Omega|\): número de elementos del espacio muestral.


## Aplicación

Se utiliza cuando **todos los resultados son igualmente probables**, por ejemplo:

- Lanzamiento de monedas.
- Lanzamiento de dados.
- Extracción de cartas de una baraja bien mezclada.

Solo puede utilizarse cuando existe equiprobabilidad entre los resultados.

```python
# MODELO LAPLACE

# Espacio muestral
omega = {1,2,3,4,5,6}

# Evento: obtener un número par
A = {2,4,6}

probabilidad = len(A) / len(omega)

print("Espacio muestral:", omega)
print("Evento:", A)
print("Probabilidad =", probabilidad)
```

# Modelo Geométrico

En el **modelo geométrico**, la probabilidad se obtiene mediante una razón de áreas (o volúmenes).

$$
\boxed{
P(A)
=
\frac{\text{Área de la región favorable}}
{\text{Área total}}.
}
$$



## Aplicaciones

- Problema del encuentro (Romeo y Julieta).
- Aguja de Buffon.
- Problemas de geometría probabilística.

La probabilidad se interpreta como la proporción del área (o volumen) ocupada por el evento de interés.

```python
import numpy as np

# MODELO GEOMÉTRICO

radio = 3
lado = 10

area_circulo = np.pi * radio**2
area_cuadrado = lado**2

probabilidad = area_circulo / area_cuadrado

print("Área del círculo:", round(area_circulo,2))
print("Área del cuadrado:", area_cuadrado)
print("Probabilidad =", round(probabilidad,4))
```

```python
import numpy as np

# Número de puntos
N = 100000

# Generamos puntos uniformes
x = np.random.uniform(0,10,N)
y = np.random.uniform(0,10,N)

# Centro del círculo
cx = 5
cy = 5
radio = 3

# Distancia al centro
dentro = (x-cx)**2 + (y-cy)**2 <= radio**2

probabilidad = np.mean(dentro)

print("Probabilidad aproximada:", probabilidad)
print("Probabilidad teórica:", round((np.pi * radio**2) / 100, 4))
```

# Modelo Frecuentista

El modelo frecuentista interpreta la probabilidad como el límite de la **frecuencia relativa** cuando un experimento se repite muchas veces bajo las mismas condiciones.

Si un evento \(A\) ocurre \(k\) veces en \(n\) repeticiones,

la frecuencia relativa es

$$
f_n(A)=\frac{k}{n}.
$$

Cuando

$$
n\rightarrow\infty,
$$

se espera que

$$
\boxed{
P(A)
=
\lim_{n\to\infty}
\frac{k}{n}.
}
$$


## Ejemplo

Al lanzar una moneda muchas veces,

$$
\frac{\text{Número de caras}}
{\text{Número de lanzamientos}}
$$

tiende a estabilizarse alrededor de

$$
\frac12.
$$

Esto es consecuencia de la **Ley de los Grandes Números**.

> **Idea clave:** La probabilidad se interpreta como la frecuencia relativa observada después de un gran número de repeticiones del experimento.

```python
#Ahora estimaremos la probabilidad de obtener cara lanzando muchas veces una moneda.
import numpy as np
import matplotlib.pyplot as plt

# MODELO FRECUENTISTA

np.random.seed(42)

N = 5000

# 0 = Sol
# 1 = Cara
lanzamientos = np.random.randint(0,2,N)

frecuencia = np.cumsum(lanzamientos) / np.arange(1,N+1)

plt.figure(figsize=(10,5))
plt.plot(frecuencia, label="Frecuencia relativa")
plt.axhline(0.5,color='red',linestyle='--',label='Probabilidad teórica')

plt.xlabel("Número de lanzamientos")
plt.ylabel("Frecuencia relativa de Cara")
plt.title("Modelo frecuentista")
plt.legend()
plt.show()
```

# Resumen de los Modelos de Probabilidad

| Modelo | Idea principal | Ejemplo |
|---------|----------------|----------|
| **Clásico (Laplace)** | Todos los resultados son equiprobables. | Dados, monedas, cartas. |
| **Geométrico** | La probabilidad es una razón de áreas o volúmenes. | Aguja de Buffon, Romeo y Julieta. |
| **Frecuentista** | La probabilidad es el límite de la frecuencia relativa. | Experimentos repetidos muchas veces. |

# Probabilidad Condicional

En muchas situaciones disponemos de **información adicional** sobre el experimento aleatorio.

Cuando sabemos que un evento ya ocurrió, la probabilidad de otros eventos puede cambiar.

A esta nueva probabilidad se le llama **probabilidad condicional**.



## Definición

Sean \(A\) y \(B\) dos eventos tales que

$$
P(B)>0.
$$

La **probabilidad condicional** de que ocurra \(A\), sabiendo que ocurrió \(B\), se define como

$$
\boxed{
P(A\,|\,B)
=
\frac{P(A\cap B)}{P(B)}.
}
$$

donde

- \(A\cap B\): ambos eventos ocurren.
- \(P(B)\): probabilidad del evento conocido.


## Interpretación

La información de que ocurrió \(B\) reduce el espacio muestral únicamente a los resultados contenidos en \(B\).

Dentro de ese nuevo espacio, calculamos qué proporción pertenece también a \(A\).

En otras palabras,

$$
P(A\,|\,B)
=
\frac{
\text{casos favorables dentro de }B
}{
\text{casos posibles dentro de }B
}.
$$

La probabilidad condicional actualiza las probabilidades cuando se dispone de información adicional.

```python
# Probabilidad Condicional
# Ejemplo con un dado

# Espacio muestral
omega = {1,2,3,4,5,6}

# Eventos
A = {2,4,6}      # Número par
B = {4,5,6}      # Número mayor que 3

# Intersección
A_inter_B = A.intersection(B)

# Probabilidades (Modelo de Laplace)
P_B = len(B)/len(omega)
P_A_inter_B = len(A_inter_B)/len(omega)

# Probabilidad condicional
P_A_dado_B = P_A_inter_B / P_B

print("Espacio muestral:", omega)
print("Evento A (par):", A)
print("Evento B (>3):", B)
print("A ∩ B:", A_inter_B)

print(f"\nP(B) = {P_B:.2f}")
print(f"P(A ∩ B) = {P_A_inter_B:.2f}")
print(f"P(A | B) = {P_A_dado_B:.2f}")
```

```python
#Ejemplo 2, la probabilidad que sea un As de corazones en una baraja inglesa
# Modelo de Laplace con cartas

total_cartas = 52
corazones = 13
ases = 4

# Solo existe un As de corazones
as_corazones = 1

P_B = corazones / total_cartas
P_A_inter_B = as_corazones / total_cartas

P_A_dado_B = P_A_inter_B / P_B

print(f"P(Corazones) = {P_B:.2f}")
print(f"P(As ∩ Corazones) = {P_A_inter_B:.4f}")
print(f"P(As | Corazones) = {P_A_dado_B:.4f}")
```

# Probabilidad Laplaciana
Para un experimento aleatorio, se tiene que $\Omega$ es el espacio muestral (conjunto de posibles resultados).

Sea $A$ un evento, entonces defimos la probabilidad de $A$ como sigue:

$$ \mathbb{P}(A) = \frac{|A|}{|\Omega|}. $$

**Ejemplo.** Sea el experimento de lanzar una moneda 3 veces, de manera independiente, es decir, el $i$-ésimo lanzamiento no afecta al $i+1$ lanzamiento. El espacio muestral es:

$$ \Omega = \{A,S\} \times \{A,S\} \times \{A,S\} = \{A,S\}^{3} , A=\{águlas\}, S=\{soles\}$$

$$ A \times B = \{(a,b) \mid a\in A, b\in B\} $$
y
$$\Omega = \{ (s,s,s),(s,a,s),(s,s,a),(s,a,a),(a,a,a),(a,s,a),(a,s,s),(a,a,s) \}$$

y se tiene que $|\Omega|= 2^3 = 8$.

Sean los eventos:
  * $A$: todos los resultados son águila
  * $S$: todos los resultados son soles
  * $E$: por lo menos un resultado es sol (1 o 2 o 3 son soles)

Vamos a calculas las probabilidades de $A,S$ y $E$.

Primero para el evento $A$, vamos a descomponer al conjunto de la siguiente manera

$$A = A_1 \cap A_2 \cap A_3,$$
donde $A_i$: cae águila en el $i$-ésimo lanzamiento.


```python
# Ejemplo Laplaciano

# Probabilidad de obtener suma = 7 al lanzar dos dados

# Espacio muestral
omega = [(i, j) for i in range(1,7) for j in range(1,7)]

# Evento: suma igual a 7
A = [(i,j) for (i,j) in omega if i+j==7]

# Probabilidad
P = len(A)/len(omega)

print("Número de casos posibles:", len(omega))
print("Casos favorables:", A)
print("Número de casos favorables:", len(A))
print(f"\nP(Suma = 7) = {P:.4f}")
```
